#### 03 Get the employees who earn more than their managers

-------------------------------------------------------------
##### SQL Query

```python
%sql

-- Find employees who earn more than their managers in the same department
SELECT 
    e.departmentid,
    e.Name as employee_name,
    e.role as employee_role,
    e.salary as employee_salary,
    m.Name as manager_name,
    m.role as manager_role,
    m.salary as manager_salary,
    e.salary - m.salary as salary_difference
FROM dev.spark_db.employee e
INNER JOIN dev.spark_db.employee m 
    ON e.departmentid = m.departmentid
    AND e.role NOT LIKE '%Manager%'
    AND m.role LIKE '%Manager%'
    AND e.salary > m.salary
WHERE e.departmentid IS NOT NULL
ORDER BY e.departmentid, salary_difference DESC
;
```

--------------------------------------------------------
##### Pyspark

```python
from pyspark.sql.functions import col, expr, max, count, min, sum, avg,year, month, dense_rank, desc, lag, countDistinct
from pyspark.sql.window import Window
from pyspark.sql.types     import StructType, StructField, StringType, IntegerType, DateType
```

######Method 1: Using DataFrame API (Recommended)
This approach is the most common and "PySpark way" to achieve the result.

```python

employee_df = spark.table('dev.spark_db.employee').alias("e")
managers_df = spark.table("dev.spark_db.employee").alias("m")

join_expr = expr("e.departmentid == m.departmentid and  e.role NOT LIKE '%Manager%' and m.role LIKE '%Manager%' and e.salary > m.salary" )

joined_df = ( employee_df.join(managers_df, join_expr, 'inner')
                         .filter( col("e.departmentid").isNotNull())
                         .withColumns({ "employee_name":     expr("e.name")   ,
                                        "employee_role":     expr("e.role")   ,
                                        "employee_salary":   expr("e.salary") ,
                                        "manager_name":      expr("m.name")   ,
                                        "manager_role":      expr("m.role")   ,
                                        "manager_salary":    expr("m.salary") ,
                                        "salary_difference": expr("e.salary - m.salary")
                                      })
                         .select("e.departmentid","employee_name", "employee_role", "employee_salary", "manager_name", "manager_role", "manager_salary", "salary_difference" ).distinct()
                         .orderBy(col("e.departmentid"), col("salary_difference").desc() )

            ) 
joined_df.display()
```

```python
# Alternative method using window functions to find max manager salary per department

from pyspark.sql.functions import when, max as spark_max

# Add employee type and get max manager salary per department
result_df_alt = (raw_emp_df
    .withColumn("is_manager", when(col("role").like("%Manager%"), 1).otherwise(0))
    .withColumn("manager_salary", 
                when(col("role").like("%Manager%"), col("salary")).otherwise(None))
    .groupBy("departmentid")
    .agg(spark_max("manager_salary").alias("max_manager_salary"))
    .join(raw_emp_df, "departmentid")
    .filter(
        (~col("role").like("%Manager%")) & 
        (col("salary") > col("max_manager_salary"))
    )
    .select(
        "employee_id",
        "Name",
        "role",
        "salary",
        "max_manager_salary",
        (col("salary") - col("max_manager_salary")).alias("salary_difference")
    )
    .orderBy("departmentid", col("salary_difference").desc())
)

result_df_alt.display()
```

#### 11 Write a query to find gaps in a sequence of numbers (missing IDs)
-------------------------------------------------------------
##### SQL Query

Let's use the employees table but we'll make a modification in records cause as it is right now there is no missing/skipping ID in the sequence.

##### Replace employee_id 2 --> 32 and 5 --> 34


```python
%sql

UPDATE dev.spark_db.employee 
SET employee_id= CASE employee_id 
                      WHEN 2 THEN 32
                      WHEN 5 THEN 34
                      ELSE employee_id
                 END
;

SELECT  * FROM  dev.spark_db.employee as c
ORDER BY employee_id
;
```

---------------------------------------------------------
This SQL query finds any gaps in a sequence of employee ID numbers (such as a missing ID like 10 when IDs 9 and 11 exist) from the table dev.spark_db.employee and lists them in order.

##### Breakdown of the Code

- FROM dev.spark_db.employee AS e1: Treats the employee table as a first set of data (e1) to look at each row one by one.
- SELECT (EMPLOYEE_ID + 1) as missing_id: Takes the ID number of the current employee, adds 1 to it, and labels this new number as missing_id.
- WHERE NOT EXISTS (...):  Keeps only the rows where the subquery returns nothing. Filters the results to keep only the numbers where a specific condition is not true.


- SELECT 1 FROM dev.spark_db.employee as e2 WHERE e2.employee_id = e1.employee_id + 1:  
        Returns the literal number 1 instead of actual column data. This optimizes performance because the query only needs to verify if a row exists, not retrieve its data. Looks at the same employee table a second time (e2) to check if the next consecutive ID (current ID + 1) actually exists in the table.
- The Logic: If current ID + 1 does not exist in the table, it means there is a gap, so the calculated missing_id is kept.
- ORDER BY missing_id;: Sorts the final list of missing ID numbers from smallest to largest.

```python
%sql
            SELECT (EMPLOYEE_ID + 1) as missing_id
            FROM   dev.spark_db.employee AS e1
            WHERE NOT EXISTS( SELECT 1 
                              FROM dev.spark_db.employee as e2
                              WHERE e2.employee_id = e1.employee_id + 1
                            ) 
            ORDER BY missing_id;
```

#### 36  Find the difference between current row's sales and previous row's sales by product

-------------------------------------------------------------
##### SQL Query

```python
%sql
SELECT order_date
      ,product_name
      ,SUM(quantity * price) as sales
      ,LAG(SUM(quantity * price)) OVER(PARTITION BY product_name ORDER BY order_date) AS PREV_SALES
      ,SUM(quantity * price) - LAG(SUM(quantity * price)) OVER(PARTITION BY product_name ORDER BY order_date) AS sales_diff
FROM   dev.spark_db.customers as c
GROUP BY order_date, product_name
ORDER BY product_name, order_date


;
```

--------------------------------------------------------
##### Pyspark

```python
from pyspark.sql.functions import col, expr, max, count, min, sum, avg,year, month, dense_rank, desc, lag, countDistinct
from pyspark.sql.window import Window
from pyspark.sql.types     import StructType, StructField, StringType, IntegerType, DateType
```

```python

grouped_date = (raw_cust_df.groupBy("order_date", "product_name")
                           .agg( expr('sum(quantity * price) as Sales') )
               )

windowSpec= Window.partitionBy("product_name").orderBy("order_date")

cust_df = ( grouped_date.withColumns({"Prev_sales": lag('Sales').over(windowSpec),
                                      "Sales_diff": expr('Sales - Prev_sales')
                                    })
                        .orderBy("product_name", "order_date")

         )

cust_df.display()       
```
