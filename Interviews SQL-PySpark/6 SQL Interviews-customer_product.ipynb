#### 50 Write a query to concatenate employee names in each department 
-------------------------------------------------------------
##### SQL Query

```python
%sql

SELECT     NAME, CONCAT( NAME, "-->", DEPARTMENTNAME) AS EMP_DEP
FROM       dev.spark_db.employee    as e
INNER JOIN dev.spark_db.department  as d on e.departmentid = d.departmentid

;
```

--------------------------------------------------------
##### Pyspark

```python
from pyspark.sql.functions import col, expr, max, count, min, sum, avg,year, month, dense_rank, desc, lag, countDistinct, concat_ws
from pyspark.sql.window import Window
from pyspark.sql.types     import StructType, StructField, StringType, IntegerType, DateType
```

######Method 1: Using DataFrame API (Recommended)
This approach is the most common and "PySpark way" to achieve the result.

```python

emps= spark.table('dev.spark_db.employee').alias('e')
dept= spark.table('dev.spark_db.department').alias('d')

join_condition = col('e.departmentid') == col('d.departmentid')

emp_df = ( emps.join(dept, join_condition, 'inner')
                     .withColumn("Emp_dept", concat_ws('-->', col("name"), col("departmentname") ) )
                     .select('name', 'd.departmentid', 'Emp_dept')
         )

emp_df.display()
```
