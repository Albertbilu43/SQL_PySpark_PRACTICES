#### 17 Write a query to caclulate the difference between current and previous salary row
-------------------------------------------------------------
##### SQL Query


```python
%sql

SELECT DEPARTMENTID, salary
      ,LAG(SALARY) OVER(PARTITION BY DEPARTMENTID ORDER BY SALARY) PREV_SALARY
      ,SALARY - LAG(SALARY) OVER(PARTITION BY DEPARTMENTID ORDER BY SALARY) SALARY_DIFF
FROM   dev.spark_db.employee as c
WHERE departmentid IS NOT NULL

;
```

--------------------------------------------------------
##### Pyspark

```python
from pyspark.sql.functions import col, expr, max, count, min, sum, avg,year, month, dense_rank, desc, lag, countDistinct, when
from pyspark.sql.window import Window
from pyspark.sql.types     import StructType, StructField, StringType, IntegerType, DateType
```

######Method 1: Using DataFrame API (Recommended)
This approach is the most common and "PySpark way" to achieve the result.

```python
windowSpec =  Window.partitionBy("departmentid").orderBy("salary")

emp_df = ( raw_emp_df.filter(col('departmentid').isNotNull())
                     .withColumns({ 'Prev_salary': lag('salary').over(windowSpec)
                                   ,'Salary_Diff': col('salary') - lag('salary').over(windowSpec)
                                  })
                     .select('departmentid', 'salary', 'Prev_salary', 'Salary_Diff')
          )

emp_df.display()
```

#### 19 Find employees with Salary greater than average salary in the entire company, order by salary descending

-------------------------------------------------------------
##### SQL Query

```python
%sql

SELECT   name, salary, (SELECT AVG(SALARY) FROM dev.spark_db.employee) as AVG_SALARY
FROM     dev.spark_db.employee as c
group by name, salary
having   salary > AVG_SALARY

;
```

--------------------------------------------------------
##### Pyspark

```python
from pyspark.sql.functions import col, expr, max, count, min, sum, avg, year, month, dense_rank, desc, lag, countDistinct
from pyspark.sql.window import Window
from pyspark.sql.types     import StructType, StructField, StringType, IntegerType, DateType
```

######Method 1: Using DataFrame API (Recommended)
This approach is the most common and "PySpark way" to achieve the result.

```python
from pyspark.sql.functions import lit

avg_sal= raw_emp_df.agg(avg('salary')).first()[0] #### avg_sal is a Python variable, not a column in the DataFrame. 
#### It doesn't exist in the DataFrame schema so to be able to use it in a dataframe YOU MUST wrap it in a lit() function.
#### The fix is to add avg_sal as a literal column using lit():

emp_df = ( raw_emp_df.filter( col('salary') > avg_sal)
                     .select( col('name'), col('salary'), lit(avg_sal).alias('avg_sal') )
          )

emp_df.display()
```

Filter to departmentid 2 to see that case particularly

```python
from pyspark.sql.functions import col, expr, max, count, min, sum, avg,year, month, dense_rank, desc, lag, countDistinct, when
from pyspark.sql.window import Window
from pyspark.sql.types     import StructType, StructField, StringType, IntegerType, DateType
```

```python

emp_df =  spark.table("dev.spark_db.employee")
windowSpec = Window.partitionBy('departmentid').orderBy(col('salary').desc())

modify_sal= (emp_df.filter(col('departmentid') == 2)
                   .withColumn('Same_sal', when(col('salary') == lag('salary').over(windowSpec), 'Yes') ) 
                   .filter(col('Same_sal') == 'Yes' )
                   .select('Name', 'salary')
            )


modify_sal.display()
```
