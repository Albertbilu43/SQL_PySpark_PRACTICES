##### Customer table 
How to use SparkSession to read a csv file 



```python
raw_cust_df= (
          spark.read.format('csv')
                    .option("header", "true")        # Use the first line as the column names
                    .option("inferSchema", "true")   # Automatically detect column data types e.i if a column contains names then infer is String
                    .load("/Volumes/dev/spark_db/datasets/spark_programmingHBL/data/customers.csv") # Read the file from this location
)
```

```python
from pyspark.sql.functions import expr, col

cust_df = raw_cust_df
#cust_df.display()

```

#### Load data into the Customers table

```python
cust_df.write.mode("overwrite").saveAsTable("dev.spark_db.delcust2")
```


#### Query the table

```python
%sql

SELECT count(*) FROM dev.spark_db.delcust2  -- count all records

--df = spark.table("dev.spark_db.delcust2")
--df.display()
```
