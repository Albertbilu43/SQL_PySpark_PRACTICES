1 Create new repository

    1 -  Go to GitHub/Repositories/
    2 -  Click on New(the green button)
       2 Name it "scholarnest_datasets"
       3 Create Repository 

2 Upload a folder in a repository

    1 - If the repository is empty, to start uploading files you first need to create a codespace.
        - Click on Create a Codespace (in the section "Start coding with Codespaces" )
        - Click on Create New Codespace 
        - Return to your Homepage of GitHub
    2 -  Go to GitHub/Repositories/
    3 -  Select the repository "scholarnest_datasets"
         Click on Add a file
         Click on Upload files
         Click on  Choose your files
         Drag and drop the desired folder (spark_programmingHBL) to the repository space, once all files are complete  click cancel that popup window 
         Click on Commit changes

         If teh folder/file already exists then it will be replaced but you can delete it as it's mentioned in 4 Deleting....
         Download the "data" file from the volume and upload it to GitHub separately
         Download the "HBL_SQL Interviews" file and upload it to GitHub separately 


3 Delete repository

    1 -  Go to GitHub/Repositories/
    2 -  Click on Settings
       1 Delete Create Repository 
       2 It will ask you to confirm this action


4 Deleting a directory/folder

    1 -  Select the directory/folder
    2 -  More Options (top-right)
    3 -  Delete Directory
    4 -  Commit changes
