### 3 Create Tables


-----------------------------
##### CUSTOMERS

```python
%sql

CREATE TABLE IF NOT EXISTS dev.spark_db.customers(
  customerid INT,
	customer_name STRING,
	order_id INT,
	productid INT,
	product_name STRING,
	quantity INT,
	price INT,
	order_date DATE	
);
```

-----------------------------
##### ORDERS

```python
%sql

CREATE TABLE IF NOT EXISTS dev.spark_db.orders(
  order_id INT,
	customerid INT,
	productid INT,
	product_name STRING,
	region STRING,
	order_date DATE	
);
```

-----------------------------
##### PRODUCTS


```python
%sql
CREATE TABLE IF NOT EXISTS dev.spark_db.products(
	productid INT,
	product_name STRING,
	unit_product_price INT	
);
```

-----------------------------
##### RETURNS

```python
%sql
CREATE TABLE IF NOT EXISTS dev.spark_db.returns(
	order_id INT,
	customerid INT,
	productid INT,
	product_name STRING,
	RETURN_DATE DATE
);
```
