#### CREATE TABLE: employee

```python
%sql

CREATE TABLE IF NOT EXISTS dev.spark_db.employee(
  employee_id INT,
	name STRING,
	departmentid INT,
	role STRING,
	salary INT,
	product_name STRING,
	startdate DATE,
	enddate DATE	
);
```

```python
"""-- Using DELTA
CREATE TABLE EMPLOYEE2 (
    employee_id INT NOT NULL,
    name STRING NOT NULL,
    departmentid INT,
    role STRING,
    salary INT,
    startdate TIMESTAMP,
    enddate TIMESTAMP,
    CONSTRAINT employee_pk PRIMARY KEY (employee_id) RELY,
    CONSTRAINT fk_department FOREIGN KEY (departmentid) REFERENCES DEPARTMENT(departmentid) NORELY
)
USING DELTA;"""
```
